```python
reviews = [
    "This product is amazing, I love it so much.",
    "The battery life is excellent and the build feels premium.",
    "I am not disappointed at all, this is wonderful.",
    "This product is terrible, I regret buying it.",
    "The battery drains fast and the build feels cheap.",
    "I am not happy with this at all, very disappointing.",
]

labels = ["Positive", "Positive", "Positive", "Negative", "Negative", "Negative"]
```

```python
from nltk import pos_tag

sentence_tokens = ["I", "went", "running", "yesterday"]

tags = pos_tag(sentence_tokens)

print(tags)
```

```python
from nltk.corpus import wordnet

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

```python
sentence_tokens = ["I", "went", "running", "yesterday", "and", "studies", "were", "better"]

pos_tags = pos_tag(sentence_tokens)

lemmatized = [lemmatizer.lemmatize(word, pos=get_wordnet_pos(tag)) for word, tag in pos_tags]

print(lemmatized)
```

```python
def preprocess(text):
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t.isalpha()]
    tokens = [t.lower() for t in tokens if t.lower() not in stop_words]
    tags = pos_tag(tokens)
    lemmatized = [lemmatizer.lemmatize(word, pos=get_wordnet_pos(tag)) for word, tag in tags]
    return " ".join(lemmatized)
```

```python
test_review = "I am not disappointed at all, the products were amazing!"
cleaned = preprocess(test_review)

print(cleaned)
```

```python
cleaned_reviews = [preprocess(review) for review in reviews]

for original, cleaned in zip(reviews, cleaned_reviews):
    print(original, "->", cleaned)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(cleaned_reviews)

print(X.shape)
print(vectorizer.get_feature_names_out())
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()

model.fit(X, labels)

predictions = model.predict(X)

for review, actual, predicted in zip(reviews, labels, predictions):
    print(actual, "|", predicted, "|", review)
```

```python
import random

random.seed(42)

nouns = ["product", "item", "phone", "laptop", "camera", "app", "service",
         "battery", "screen", "design", "delivery", "sound", "build"]

pos_adj = ["amazing", "excellent", "wonderful", "fantastic", "impressive",
           "reliable", "comfortable", "premium", "smooth", "outstanding"]

neg_adj = ["terrible", "awful", "disappointing", "unreliable",
           "uncomfortable", "sluggish", "cheap", "defective", "poor", "broken"]

pos_templates = [
    "The {noun} is {adj}, I am very happy with it.",
    "This {noun} works great, truly {adj} experience overall.",
    "I love how {adj} the {noun} feels, highly recommend it.",
    "The {noun} is not bad at all, actually quite {adj}.",
]

neg_templates = [
    "The {noun} is {adj}, I regret buying it.",
    "This {noun} barely works, truly {adj} experience overall.",
    "I hate how {adj} the {noun} feels, do not recommend it.",
    "The {noun} is not good at all, actually quite {adj}.",
]

def generate_reviews(templates, adjectives, n):
    reviews = set()
    while len(reviews) < n:
        t = random.choice(templates)
        reviews.add(t.format(noun=random.choice(nouns), adj=random.choice(adjectives)))
    return list(reviews)

positive_reviews = generate_reviews(pos_templates, pos_adj, 60)
negative_reviews = generate_reviews(neg_templates, neg_adj, 60)

reviews = positive_reviews + negative_reviews
labels = ["Positive"] * len(positive_reviews) + ["Negative"] * len(negative_reviews)

print(len(reviews), "reviews total")
print(reviews[0])
```

```python
cleaned_reviews = [preprocess(review) for review in reviews]

print(reviews[0], "->", cleaned_reviews[0])
print(reviews[70], "->", cleaned_reviews[70])
```

```python
from sklearn.model_selection import train_test_split

X_train_text, X_test_text, y_train, y_test = train_test_split(
    cleaned_reviews, labels, test_size=0.25, random_state=42, stratify=labels
)

print("Training reviews:", len(X_train_text))
print("Testing reviews:", len(X_test_text))
```

```python
vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=2000, min_df=2)

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

model = MultinomialNB(alpha=1.0)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)
print("Accuracy:", accuracy)

print(classification_report(y_test, predictions))

print("Confusion Matrix:")
print(confusion_matrix(y_test, predictions))
```

```python
def predict_sentiment(review):
    cleaned = preprocess(review)
    vector = vectorizer.transform([cleaned])
    return model.predict(vector)[0]

new_reviews = [
    "I am not happy with this purchase, it broke fast.",
    "This is not just good, it's the best thing I've bought all year.",
    "Terrible packaging but the product itself works great.",
    "Never buying this again, complete waste of money.",
]

for review in new_reviews:
    print(predict_sentiment(review), "|", review)
```

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train, y_train)

svm_predictions = svm_model.predict(X_test)
svm_accuracy = accuracy_score(y_test, svm_predictions)

print("Naive Bayes Accuracy:", accuracy)
print("SVM Accuracy:", svm_accuracy)
```
