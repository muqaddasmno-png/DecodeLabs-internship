```python
# Create customer-level features

customer_data = df.groupby('CustomerID').agg(
    TotalOrders=('OrderID', 'count'),
    TotalQuantity=('Quantity', 'sum'),
    TotalSpending=('TotalPrice', 'sum'),
    AvgOrderValue=('TotalPrice', 'mean'),
    AvgItemsInCart=('ItemsInCart', 'mean')
).reset_index()

customer_data.head()
```

```python
customer_data.shape
```

```python
# Select features for customer clustering

features = [
    'TotalOrders',
    'TotalQuantity',
    'TotalSpending',
    'AvgOrderValue',
    'AvgItemsInCart'
]

X = customer_data[features]

X.head()
```

```python
X.describe()
```

```python
from sklearn.preprocessing import StandardScaler

# Scale the clustering features
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easy viewing
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

X_scaled_df.head()
```

```python
X_scaled_df.describe()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(range(2, 11))
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different K Values')
plt.xticks(range(2, 11))
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in [2, 3]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=300,
        random_state=42
    )

    print(f"K={k}: Silhouette Score = {score:.4f}")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Load dataset
df = pd.read_excel("Dataset for Data Analytics (1).xlsx")

# Fill missing values
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')

# Create customer-level data
customer_data = df.groupby('CustomerID').agg(
    TotalOrders=('OrderID', 'count'),
    TotalQuantity=('Quantity', 'sum'),
    TotalSpending=('TotalPrice', 'sum'),
    AvgOrderValue=('TotalPrice', 'mean'),
    AvgItemsInCart=('ItemsInCart', 'mean')
).reset_index()

# Select clustering features
features = [
    'TotalOrders',
    'TotalQuantity',
    'TotalSpending',
    'AvgOrderValue',
    'AvgItemsInCart'
]

X = customer_data[features]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Compare K=2 and K=3
for k in [2, 3]:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=300,
        random_state=42
    )

    print(f"K={k}: Silhouette Score = {score:.4f}")
```

```python
from sklearn.cluster import KMeans

final_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

customer_data['Cluster'] = final_kmeans.fit_predict(X_scaled)

customer_data.head()
```

```python
customer_data['Cluster'].value_counts().sort_index()
```

```python
cluster_summary = customer_data.groupby('Cluster')[features].mean()
cluster_summary
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=customer_data['Cluster'],
    cmap='viridis'
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Customer Clusters (PCA)')
plt.show()
```

```python
cluster_percentage = (
    customer_data['Cluster']
    .value_counts(normalize=True)
    .sort_index() * 100
)

cluster_percentage.round(2)
```
